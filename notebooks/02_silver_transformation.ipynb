```python
# Read Bronze users table
users_df = spark.table("workspace.zomato_bronze.raw_users")

# Inspect schema and sample data
users_df.printSchema()
display(users_df.limit(10))
```

```python
from pyspark.sql.functions import col, trim, lower

users_silver_df = (
    users_df
    .withColumn("name", trim(col("name")))
    .withColumn("email", lower(trim(col("email"))))
    .withColumn("gender", trim(col("gender")))
    .withColumn("marital_status", trim(col("marital_status")))
    .withColumn("occupation", trim(col("occupation")))
    .withColumn("monthly_income", trim(col("monthly_income")))
    .withColumn(
        "educational_qualifications",
        trim(col("educational_qualifications"))
    )
    .filter(col("user_id").isNotNull())
    .dropDuplicates(["user_id"])
)

display(users_silver_df.limit(10))
```

```python
users_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.users")
```

```python
restaurants_silver_df = (
    restaurants_df
    .withColumn("id", trim(col("id")))
    .withColumn("name", trim(col("name")))
    .withColumn("city", trim(col("city")))
    .withColumn("rating", trim(col("rating")))
    .withColumn("rating_count", trim(col("rating_count")))
    .withColumn("cost", trim(col("cost")))
    .withColumn("cuisine", trim(col("cuisine")))
    .withColumn("lic_no", trim(col("lic_no")))
    .withColumn("link", trim(col("link")))
    .withColumn("address", trim(col("address")))
    .withColumn("menu", trim(col("menu")))
    .filter(col("id").isNotNull())
    .dropDuplicates(["id"])
)

display(restaurants_silver_df.limit(10))

restaurants_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.restaurants")
```

```python
menus_silver_df = (
    menus_df
    .withColumn("menu_id", trim(col("menu_id")))
    .withColumn("cuisine", trim(col("cuisine")))
    .withColumn("price", trim(col("price")))
    .filter(col("menu_id").isNotNull())
    .filter(col("r_id").isNotNull())
    .filter(col("f_id").isNotNull())
    .dropDuplicates(["r_id", "f_id"])
)

display(menus_silver_df.limit(20))
```

```python
menus_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.menus")
```

```python
foods_silver_df = (
    foods_df
    .withColumn("f_id", trim(col("f_id")))
    .withColumn("item", trim(col("item")))
    .withColumn("veg_or_non_veg", trim(col("veg_or_non_veg")))
    .filter(col("f_id").isNotNull())
    .dropDuplicates(["f_id"])
)

display(foods_silver_df.limit(10))

foods_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.foods")
```

```python
from pyspark.sql.functions import col, trim
orders_silver_df = (
    orders_df
    .withColumn("order_date", col("order_date").cast("date"))
    .withColumn("sales_qty", col("sales_qty").cast("integer"))
    .withColumn("sales_amount", col("sales_amount").cast("double"))
    .withColumn("currency", trim(col("currency")))
    .withColumn("user_id", col("user_id").cast("integer"))
    .withColumn("r_id", col("r_id").cast("double"))
    .filter(
        col("order_date").isNotNull()
        & col("sales_qty").isNotNull()
        & col("sales_amount").isNotNull()
        & col("currency").isNotNull()
        & col("user_id").isNotNull()
        & col("r_id").isNotNull()
    )
    .dropDuplicates(["order_date", "user_id", "r_id"])
)

display(orders_silver_df.limit(10))

orders_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.orders")
```

```python
countries_silver_df = (
    countries_df
    .withColumn("num_code", col("num_code").cast("integer"))
    .withColumn("alpha_2_code", trim(col("alpha_2_code")))
    .withColumn("alpha_3_code", trim(col("alpha_3_code")))
    .withColumn("en_short_name", trim(col("en_short_name")))
    .withColumn("nationality", trim(col("nationality")))
    .filter(col("num_code").isNotNull())
    .dropDuplicates(["num_code"])
)

display(countries_silver_df.limit(10))

countries_silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.zomato_silver.countries")
```

```python
from pyspark.sql.functions import col, expr

orders_df = (
    spark.table("workspace.zomato_silver.orders")
    .withColumn("restaurant_id", col("r_id").cast("long"))
)

restaurants_df = (
    spark.table("workspace.zomato_silver.restaurants")
    .withColumn(
        "restaurant_id",
        expr("try_cast(id AS BIGINT)")
    )
    .filter(col("restaurant_id").isNotNull())
)

orders_restaurant_check = (
    orders_df
    .join(
        restaurants_df.select("restaurant_id"),
        "restaurant_id",
        "left"
    )
    .filter(restaurants_df.restaurant_id.isNull())
)

display(orders_restaurant_check)
```

```python
print(
    "Orders without matching restaurant:",
    orders_restaurant_check.count()
)
```

```python
menus_df = (
    spark.table("workspace.zomato_silver.menus")
    .withColumn("restaurant_id", col("r_id").cast("long"))
)

restaurants_df = (
    spark.table("workspace.zomato_silver.restaurants")
    .withColumn(
        "restaurant_id",
        expr("try_cast(id AS BIGINT)")
    )
    .filter(col("restaurant_id").isNotNull())
)

menus_restaurant_check = (
    menus_df
    .join(
        restaurants_df.select("restaurant_id"),
        "restaurant_id",
        "left"
    )
    .filter(restaurants_df.restaurant_id.isNull())
)

display(menus_restaurant_check)

print(
    "Menus without matching restaurant:",
    menus_restaurant_check.count()
)
```

```python
menus_df = spark.table("workspace.zomato_silver.menus")
foods_df = spark.table("workspace.zomato_silver.foods")

menus_food_check = (
    menus_df
    .join(
        foods_df.select("f_id"),
        "f_id",
        "left"
    )
    .filter(foods_df.f_id.isNull())
)

display(menus_food_check)

print(
    "Menus without matching food:",
    menus_food_check.count()
)
```
